```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Carga desde un archivo .csv sin indice
data = pd.read_csv('Citas_Digital_Filtrado.csv')
data.head(5)
```

```python
data.info()
```

```python

data2 = data.drop(['Fecha','Telefono ','Unnamed: 15'], axis=1)
data2
```

```python
valores_nulos = data2.isnull().sum()
valores_nulos
```

```python
#Realizamos una copia del dataframe
data3 = data2.copy()
```

```python
data3["Venta"] = data3["Venta"].fillna(0)
```

```python
data3["SDC"] = data3["SDC"].fillna(round(data2["SDC"].median(),1))
data3["PDM"] = data3["PDM"].fillna(round(data2["PDM"].mean(),1))
```

```python
valores_nulos = data3.isnull().sum()
valores_nulos
```

```python
data3["¿Por que no ha visitado la agencia?"] = data3["¿Por que no ha visitado la agencia?"].fillna("No aplica")
data3["¿Por que no hay solicitud de credito?"] = data3["¿Por que no hay solicitud de credito?"].fillna("No aplica")
data3["¿Por que no hubo prueba de manejo ?"] = data3["¿Por que no hubo prueba de manejo ?"].fillna("No aplica")
data3["¿Por que no hubo proceso wow?"] = data3["¿Por que no hubo proceso wow?"].fillna("No aplica")
data3["¿Por que se asigno antes de vistar piso?"] = data3["¿Por que se asigno antes de vistar piso?"].fillna("No aplica")
```

```python
valores_nulos = data3.isnull().sum()
valores_nulos
```

```python
data3 = data3.dropna(subset=['BDC ','Nombre Cliente ','Estatus de Lead','Potencial de compra','Asesor Asignado'])
data3
```

```python
valores_nulos = data3.isnull().sum().sum()
valores_nulos
```
